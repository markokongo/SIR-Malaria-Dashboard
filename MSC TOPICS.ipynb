2. Important mathematical models
A. Exponential growth model
When nutrients are abundant and environmental conditions remain favourable, microbial populations may grow approximately exponentially:
\[
\frac{dN}{dt}=rN
\]

where:
- \(N(t)\) is the microbial population at time \(t\).
- \(r\) is the net per-capita growth rate.
- \(t\) is time.
The solution is
\[
N(t)=N_0e^{rt},
\]

where \(N_0\) is the initial population.

This model is useful for studying early bacterial growth, but it does not account for nutrient depletion or overcrowding.
B. Logistic growth model
When resources are limited, a logistic model is often more realistic:
\[
\frac{dN}{dt}=rN\left(1-\frac{N}{K}\right),
\]

where \(K\) is the environmental carrying capacity.

As the population approaches \(K\), its growth slows.
